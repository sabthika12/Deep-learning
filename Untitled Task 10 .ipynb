```python
# Step 1: Import the required libraries
# Code cell 3
# Cell 3 of 28

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
import time
import random
import numpy as np

print("PyTorch version:", torch.__version__)

# Step 2: Set random seeds and select the device
# Cell 4 of 28
# Step 2: Set random seeds and select the device
# Code cell 5
# Cell 5 of 28

torch.manual_seed(42)
np.random.seed(42)
random.seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# Step 3: Load and prepare the MNIST dataset
# Cell 6 of 28
# Step 3: Load and prepare the MNIST dataset
# Code cell 7
# Cell 7 of 28

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,))
])

train_dataset = datasets.MNIST(
    root="./data", train=True, download=True, transform=transform
)
test_dataset = datasets.MNIST(
    root="./data", train=False, download=True, transform=transform
)

train_loader = DataLoader(train_dataset, batch_size=128, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=256, shuffle=False)

print("Training images:", len(train_dataset))
print("Testing images:", len(test_dataset))

# Step 4: Display sample images
# Cell 8 of 28
# Step 4: Display sample images
# Code cell 9
# Cell 9 of 28

images, labels = next(iter(train_loader))

fig, axes = plt.subplots(1, 5, figsize=(10, 2))
for i in range(5):
    axes[i].imshow(images[i].squeeze(), cmap="gray")
    axes[i].set_title(f"Label: {labels[i].item()}")
    axes[i].axis("off")
plt.tight_layout()
plt.show()

# Step 5: Define the neural network model
# Cell 10 of 28
# Step 5: Define the neural network model
# Code cell 11
# Cell 11 of 28

class DigitClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        self.network = nn.Sequential(
            nn.Flatten(),
            nn.Linear(28 * 28, 256),
            nn.ReLU(),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Linear(128, 10)
        )

    def forward(self, x):
        return self.network(x)

print("Neural network model defined.")

# Step 6: Define a function to evaluate the model
# Cell 12 of 28
# Step 6: Define a function to evaluate the model
# Code cell 13
# Cell 13 of 28

def evaluate(model, loader):
    model.eval()
    correct = 0
    total = 0
    total_loss = 0.0
    criterion = nn.CrossEntropyLoss()

    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            outputs = model(x)
            loss = criterion(outputs, y)
            total_loss += loss.item() * y.size(0)
            correct += (outputs.argmax(dim=1) == y).sum().item()
            total += y.size(0)

    return total_loss / total, 100.0 * correct / total

# Step 7: Train a baseline model
# Cell 14 of 28
# Step 7: Train a baseline model
# Code cell 15
# Cell 15 of 28

baseline_model = DigitClassifier().to(device)
criterion = nn.CrossEntropyLoss()
baseline_optimizer = optim.SGD(baseline_model.parameters(), lr=0.01)

baseline_epochs = 3
baseline_losses = []
start_time = time.time()

for epoch in range(baseline_epochs):
    baseline_model.train()
    running_loss = 0.0

    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        baseline_optimizer.zero_grad()
        outputs = baseline_model(x)
        loss = criterion(outputs, y)
        loss.backward()
        baseline_optimizer.step()
        running_loss += loss.item()

    avg_loss = running_loss / len(train_loader)
    baseline_losses.append(avg_loss)
    _, test_acc = evaluate(baseline_model, test_loader)
    print(f"Baseline Epoch {epoch + 1}/{baseline_epochs} | "
          f"Loss: {avg_loss:.4f} | Test accuracy: {test_acc:.2f}%")

baseline_time = time.time() - start_time
baseline_test_loss, baseline_test_acc = evaluate(baseline_model, test_loader)
print(f"Baseline training time: {baseline_time:.1f} seconds")
print(f"Baseline final test accuracy: {baseline_test_acc:.2f}%")

# Step 8: Create a fresh model for optimized training
# Cell 16 of 28
# Step 8: Create a fresh model for optimized training
# Code cell 17
# Cell 17 of 28

optimized_model = DigitClassifier().to(device)
criterion = nn.CrossEntropyLoss()

# Adam adapts learning rates for model parameters.
optimized_optimizer = optim.Adam(optimized_model.parameters(), lr=0.001)

# Reduce the learning rate when validation performance stops improving.
scheduler = optim.lr_scheduler.StepLR(
    optimized_optimizer, step_size=2, gamma=0.5
)

optimized_epochs = 3
optimized_losses = []
optimized_accuracies = []
start_time = time.time()

print("Optimized model and training settings are ready.")

# Step 9: Train using optimization techniques
# Cell 18 of 28
# Step 9: Train using optimization techniques
# Code cell 19
# Cell 19 of 28

for epoch in range(optimized_epochs):
    optimized_model.train()
    running_loss = 0.0

    for x, y in train_loader:
        x, y = x.to(device), y.to(device)

        optimized_optimizer.zero_grad()
        outputs = optimized_model(x)
        loss = criterion(outputs, y)
        loss.backward()

        # Gradient clipping helps prevent unusually large gradients.
        torch.nn.utils.clip_grad_norm_(optimized_model.parameters(), max_norm=1.0)

        optimized_optimizer.step()
        running_loss += loss.item()

    scheduler.step()
    avg_loss = running_loss / len(train_loader)
    optimized_losses.append(avg_loss)
    _, test_acc = evaluate(optimized_model, test_loader)
    optimized_accuracies.append(test_acc)

    current_lr = optimized_optimizer.param_groups[0] ["lr"]
    print(f"Optimized Epoch {epoch + 1}/{optimized_epochs} | "
          f"Loss: {avg_loss:.4f} | Test accuracy: {test_acc:.2f}% | "
          f"Learning rate: {current_lr:.5f}")

optimized_time = time.time() - start_time
optimized_test_loss, optimized_test_acc = evaluate(optimized_model, test_loader)
print(f"Optimized training time: {optimized_time:.1f} seconds")
print(f"Optimized final test accuracy: {optimized_test_acc:.2f}%")

# Step 10: Plot training loss
# Cell 20 of 28
# Step 10: Plot training loss
# Code cell 21
# Cell 21 of 28

plt.figure(figsize=(8, 5))
plt.plot(range(1, baseline_epochs + 1), baseline_losses, marker="o",
         label="Baseline: SGD")
plt.plot(range(1, optimized_epochs + 1), optimized_losses, marker="s",
         label="Optimized: Adam + scheduler + clipping")
plt.xlabel("Epoch")
plt.ylabel("Average training loss")
plt.title("Training Loss Comparison")
plt.legend()
plt.grid(True)
plt.show()

# Step 11: Compare final model performance
# Cell 22 of 28
# Step 11: Compare final model performance
# Code cell 23
# Cell 23 of 28

print("Performance comparison")
print("-" * 35)
print(f"Baseline test accuracy : {baseline_test_acc:.2f}%")
print(f"Optimized test accuracy: {optimized_test_acc:.2f}%")
print(f"Baseline training time : {baseline_time:.1f} seconds")
print(f"Optimized training time: {optimized_time:.1f} seconds")
print(f"Accuracy change        : {optimized_test_acc - baseline_test_acc:+.2f} percentage points")

if optimized_test_acc > baseline_test_acc:
    print("The optimized setup achieved higher test accuracy in this run.")
elif optimized_test_acc < baseline_test_acc:
    print("The baseline achieved higher test accuracy in this run.")
else:
    print("Both setups achieved the same test accuracy in this run.")

print("Note: results and training time can vary by hardware and run.")

# Step 12: Test the optimized model on sample images
# Cell 24 of 28
# Step 12: Test the optimized model on sample images
# Code cell 25
# Cell 25 of 28

optimized_model.eval()
sample_images, sample_labels = next(iter(test_loader))
sample_images_device = sample_images[:8].to(device)

with torch.no_grad():
    predictions = optimized_model(sample_images_device).argmax(dim=1).cpu()

fig, axes = plt.subplots(2, 4, figsize=(10, 5))
for i, ax in enumerate(axes.flat):
    ax.imshow(sample_images[i].squeeze(), cmap="gray")
    ax.set_title(f"Actual: {sample_labels[i].item()}\nPredicted: {predictions[i].item()}")
    ax.axis("off")
plt.tight_layout()
plt.show()

# Step 13: Result
# Cell 26 of 28
# Step 13: Result
# Code cell 27
# Cell 27 of 28

print("Optimization techniques were applied using PyTorch.")
print("Compared SGD baseline training with Adam, learning-rate scheduling,")
print("and gradient clipping using test accuracy and training loss.")
```
