```python
# Step 1 — Import the required libraries

# Code cell 3
# Cell 3 of 26

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
import matplotlib.pyplot as plt

print("PyTorch version:", torch.__version__)

# Step 2 — Select the device
# Cell 4 of 26

# Step 2 — Select the device
# Use GPU if available; otherwise use CPU.

# Code cell 5
# Cell 5 of 26

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# Step 3 — Load the MNIST dataset
# Cell 6 of 26

# Step 3 — Load the MNIST dataset
# MNIST contains 28 × 28 grayscale handwritten digit images. ToTensor() converts pixel values to the range 0–1.

# Code cell 7
# Cell 7 of 26

transform = transforms.ToTensor()

train_dataset = datasets.MNIST("./data", train=True, download=True, transform=transform)
test_dataset = datasets.MNIST("./data", train=False, download=True, transform=transform)

train_loader = DataLoader(train_dataset, batch_size=128, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=128, shuffle=False)

print("Training images:", len(train_dataset))
print("Testing images:", len(test_dataset))

# Step 4 — Display a sample image
# Cell 8 of 26

# Step 4 — Display a sample image

# Code cell 9
# Cell 9 of 26

sample_image, sample_label = train_dataset[0]
plt.imshow(sample_image.squeeze(), cmap="gray")
plt.title(f"Sample MNIST Image (Label: {sample_label})")
plt.axis("off")
plt.show()

# Step 5 — Define the autoencoder model
# Cell 10 of 26

# Step 5 — Define the autoencoder model
# The encoder compresses 784 pixels to 32 values. The decoder reconstructs the 784 pixels.

# Code cell 11
# Cell 11 of 26

class Autoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(28 * 28, 128),
            nn.ReLU(),
            nn.Linear(128, 32),
            nn.ReLU()
        )
        self.decoder = nn.Sequential(
            nn.Linear(32, 128),
            nn.ReLU(),
            nn.Linear(128, 28 * 28),
            nn.Sigmoid()
        )

    def forward(self, x):
        x = x.view(x.size(0), -1)
        encoded = self.encoder(x)
        decoded = self.decoder(encoded)
        return decoded

model = Autoencoder().to(device)
print(model)

# Step 6 — Define the loss function and optimizer
# Cell 12 of 26

# Step 6 — Define the loss function and optimizer
# MSE measures the difference between original and reconstructed pixels. Adam updates model parameters.

# Code cell 13
# Cell 13 of 26

criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Step 7 — Set training settings
# Cell 14 of 26

# Step 7 — Set training settings

# Code cell 15
# Cell 15 of 26

num_epochs = 5
loss_history = []

# Step 8 — Train the autoencoder
# Cell 16 of 26

# Step 8 — Train the autoencoder
# Each image is both the input and the target, so the model learns to reconstruct the same image.

# Code cell 17
# Cell 17 of 26

for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0

    for images, _ in train_loader:
        images = images.to(device)
        outputs = model(images)
        targets = images.view(images.size(0), -1)
        loss = criterion(outputs, targets)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * images.size(0)

    epoch_loss = running_loss / len(train_dataset)
    loss_history.append(epoch_loss)
    print(f"Epoch [{epoch + 1}/{num_epochs}], Loss: {epoch_loss:.6f}")

# Step 9 — Plot the training loss
# Cell 18 of 26

# Step 9 — Plot the training loss

# Code cell 19
# Cell 19 of 26

plt.figure(figsize=(8, 5))
plt.plot(range(1, num_epochs + 1), loss_history, marker="o")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.title("Autoencoder Training Loss")
plt.grid(True)
plt.show()

# Step 10 — Reconstruct test images
# Cell 20 of 26

# Step 10 — Reconstruct test images

# Code cell 21
# Cell 21 of 26

model.eval()
with torch.no_grad():
    test_images, _ = next(iter(test_loader))
    test_images_device = test_images.to(device)
    reconstructed = model(test_images_device).cpu()

print("Original image batch shape:", test_images.shape)
print("Reconstructed batch shape:", reconstructed.shape)

# Step 11 — Display original and reconstructed images
# Cell 22 of 26

# Step 11 — Display original and reconstructed images
# Top row: original images. Bottom row: reconstructed images.

# Code cell 23
# Cell 23 of 26

number_of_images = 8
plt.figure(figsize=(14, 4))

for i in range(number_of_images):
    plt.subplot(2, number_of_images, i + 1)
    plt.imshow(test_images[i].squeeze(), cmap="gray")
    plt.title("Original")
    plt.axis("off")

    plt.subplot(2, number_of_images, number_of_images + i + 1)
    plt.imshow(reconstructed[i].view(28, 28), cmap="gray")
    plt.title("Reconstructed")
    plt.axis("off")

plt.tight_layout()
plt.show()

# Step 12 — Calculate reconstruction error
# Cell 24 of 26

# Step 12 — Calculate reconstruction error

# Code cell 25
# Cell 25 of 26

with torch.no_grad():
    original_flat = test_images_device.view(test_images_device.size(0), -1)
    test_loss = criterion(model(test_images_device), original_flat).item()

print(f"Test reconstruction MSE: {test_loss:.6f}")
```
