```python
# Step 1 — Import the required libraries
# Code cell 3
# Cell 3 of 26

import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

# Step 2 — Select the device
# Cell 4 of 26
# Use a GPU when available; otherwise, the notebook runs on the CPU.
# Code cell 5
# Cell 5 of 26

def get_device():
    return torch.device("cuda" if torch.cuda.is_available() else "cpu")

device = get_device()
print("Using device:", device)

# Step 3 — Prepare the sample text dataset
# Cell 6 of 26
# The model learns from short sentences. You can replace these examples with your own text.
# Code cell 7
# Cell 7 of 26

sentences = [
    "deep learning is useful",
    "rnn learns word sequence",
    "word embedding represents words",
    "deep learning uses neural networks",
    "rnn processes text sequence",
    "word vectors capture meaning",
    "neural networks learn patterns",
    "embedding converts words to vectors"
]

print("Number of sentences:", len(sentences))

# Step 4 — Tokenize the text and build the vocabulary
# Cell 8 of 26
# Each unique word is assigned an integer ID. These IDs are used to look up trainable word embeddings.
# Code cell 9
# Cell 9 of 26

tokenized_sentences = [sentence.lower().split() for sentence in sentences]

vocab = sorted(set(word for sentence in tokenized_sentences for word in sentence))
word_to_idx = {word: i for i, word in enumerate(vocab)}
idx_to_word = {i: word for word, i in word_to_idx.items()}

vocab_size = len(vocab)

print("Vocabulary size:", vocab_size)
print("Vocabulary:", vocab)

# Step 5 — Convert sentences into input sequences
# Cell 10 of 26
# For this simple next-word task, each word is an input and the following word is its target.
# Code cell 11
# Cell 11 of 26

sequence_pairs = []

for sentence in tokenized_sentences:
    ids = [word_to_idx[word] for word in sentence]
    for i in range(len(ids) - 1):
        sequence_pairs.append((ids[i], ids[i + 1]))

X = torch.tensor([pair[0] for pair in sequence_pairs], dtype=torch.long)
y = torch.tensor([pair[1] for pair in sequence_pairs], dtype=torch.long)

print("Training pairs:", len(X))
print("First input word:", idx_to_word[X[0].item()])
print("First target word:", idx_to_word[y[0].item()])

# Step 6 — Define the Word Embedding + RNN model
# Cell 12 of 26
# The Embedding layer converts word IDs into dense vectors. The RNN processes the sequence of vectors, and the linear layer predicts the next word.
# Code cell 13
# Cell 13 of 26

embedding_dim = 16
hidden_dim = 32

class WordEmbeddingRNN(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        self.rnn = nn.RNN(
            input_size=embedding_dim,
            hidden_size=hidden_dim,
            batch_first=True
        )
        self.fc = nn.Linear(hidden_dim, vocab_size)

    def forward(self, x):
        embedded = self.embedding(x)
        output, hidden = self.rnn(embedded)
        logits = self.fc(output[:, -1, :])
        return logits

model = WordEmbeddingRNN(vocab_size, embedding_dim, hidden_dim).to(device)
print(model)

# Step 7 — Define the loss function and optimizer
# Cell 14 of 26
# Cross Entropy Loss is used to predict the next word from the vocabulary.
# Code cell 15
# Cell 15 of 26

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

# Step 8 — Train the RNN model
# Cell 16 of 26
# The model learns from the word pairs over several epochs.
# Code cell 17
# Cell 17 of 26

epochs = 100
loss_history = []

X_train = X.to(device).view(-1, 1)
y_train = y.to(device)

for epoch in range(epochs):
    model.train()
    optimizer.zero_grad()

    outputs = model(X_train)
    loss = criterion(outputs, y_train)

    loss.backward()
    optimizer.step()

    loss_history.append(loss.item())

    if (epoch + 1) % 10 == 0:
        print(f"Epoch [{epoch + 1}/{epochs}], Loss: {loss.item():.4f}")

# Step 9 — Plot the training loss
# Cell 18 of 26
# The graph shows how the training loss changes as the model learns the sample word sequences.
# Code cell 19
# Cell 19 of 26

plt.figure(figsize=(8, 4))
plt.plot(range(1, epochs + 1), loss_history)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Word Embedding RNN Training Loss")
plt.grid(True)
plt.show()

# Step 10 — Inspect learned word embeddings
# Cell 20 of 26
# Each word now has a learned numeric vector of length embedding_dim.
# Code cell 21
# Cell 21 of 26

embedding_weights = model.embedding.weight.detach().cpu()

print("Embedding matrix shape:", tuple(embedding_weights.shape))
for word in vocab[:10]:
    vector = embedding_weights[word_to_idx[word]].numpy()
    print(f"{word}: {vector.round(3)}")

# Step 11 — Predict the next word
# Cell 22 of 26
# This is a simple demonstration of how the trained RNN uses the learned word representations.
# Code cell 23
# Cell 23 of 26

def predict_next_word(word):
    word = word.lower()
    if word not in word_to_idx:
        print("Word not in vocabulary:", word)
        return

    model.eval()
    input_tensor = torch.tensor([[word_to_idx[word]]], dtype=torch.long, device=device)

    with torch.no_grad():
        logits = model(input_tensor)
        predicted_id = torch.argmax(logits, dim=1).item()

    print(f"Input word: {word}")
    print(f"Predicted next word: {idx_to_word[predicted_id]}")

predict_next_word("deep")
predict_next_word("rnn")

# Step 12 — Display embedding vectors in a simple chart
# Cell 24 of 26
# The chart shows the first two dimensions of each learned embedding. These two dimensions are only a small view of the full vector.
# Code cell 25
# Cell 25 of 26

plt.figure(figsize=(10, 6))
for i, word in enumerate(vocab):
    x_value = embedding_weights[i, 0].item()
    y_value = embedding_weights[i, 1].item()
    plt.scatter(x_value, y_value)
    plt.annotate(word, (x_value, y_value), fontsize=9)

plt.xlabel("Embedding dimension 1")
plt.ylabel("Embedding dimension 2")
plt.title("Word Embeddings (First Two Dimensions)")
plt.grid(True)
plt.show()
```
