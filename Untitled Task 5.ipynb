```python
# Code cell 1
# Cell 1 of 30
import torch
import torchvision
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
import numpy as np

# Step 2 — Select the device
# Cell 2 of 30
# Step 2 — Select the device
# Use a GPU if Google Colab provides one. In Colab, select Runtime → Change runtime type → T4 GPU if available.

# Code cell 3
# Cell 3 of 30
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# Step 3 — Define image preprocessing
# Cell 4 of 30
# Step 3 — Define image preprocessing
# MNIST images are 28 × 28 grayscale images. Normalize pixel values to the range [-1, 1], which matches the Generator's Tanh output.

# Code cell 5
# Cell 5 of 30
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

# Step 4 — Download the MNIST dataset
# Cell 6 of 30
# Step 4 — Download the MNIST dataset
# The dataset contains handwritten digit images from 0 to 9.

# Code cell 7
# Cell 7 of 30
train_dataset = datasets.MNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

print("Training images:", len(train_dataset))

# Step 5 — Create the DataLoader
# Cell 8 of 30
# Step 5 — Create the DataLoader
# The DataLoader provides shuffled batches of images for GAN training.

# Code cell 9
# Cell 9 of 30
batch_size = 128

train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)

# Step 6 — Display some real images
# Cell 10 of 30
# Step 6 — Display some real images
# Check a few images from the training dataset before training.

# Code cell 11
# Cell 11 of 30
real_images, labels = next(iter(train_loader))

plt.figure(figsize=(8, 4))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(real_images[i].squeeze(), cmap="gray")
    plt.title(f"Label: {labels[i].item()}")
    plt.axis("off")
plt.tight_layout()
plt.show()

# Step 7 — Define the Generator
# Cell 12 of 30
# Step 7 — Define the Generator
# The Generator takes a 100-dimensional random noise vector and creates a 28 × 28 image.

# Code cell 13
# Cell 13 of 30
latent_dim = 100

class Generator(nn.Module):
    def __init__(self):
        super().__init__()
        self.model = nn.Sequential(
            nn.Linear(latent_dim, 256),
            nn.ReLU(),
            nn.Linear(256, 512),
            nn.ReLU(),
            nn.Linear(512, 784),
            nn.Tanh()
        )

    def forward(self, z):
        return self.model(z).view(-1, 1, 28, 28)

generator = Generator().to(device)
print(generator)

# Step 8 — Define the Discriminator
# Cell 14 of 30
# Step 8 — Define the Discriminator
# The Discriminator receives an image and predicts whether it is real or generated.

# Code cell 15
# Cell 15 of 30
class Discriminator(nn.Module):
    def __init__(self):
        super().__init__()
        self.model = nn.Sequential(
            nn.Linear(784, 512),
            nn.LeakyReLU(0.2),
            nn.Linear(512, 256),
            nn.LeakyReLU(0.2),
            nn.Linear(256, 1),
            nn.Sigmoid()
        )

    def forward(self, image):
        image = image.view(image.size(0), -1)
        return self.model(image)

discriminator = Discriminator().to(device)
print(discriminator)

# Step 9 — Define the loss function and optimizers
# Cell 16 of 30
# Step 9 — Define the loss function and optimizers
# Binary Cross Entropy measures real/fake classification loss. Adam updates the Generator and Discriminator separately.

# Code cell 17
# Cell 17 of 30
criterion = nn.BCELoss()

optimizer_G = optim.Adam(generator.parameters(), lr=0.0002, betas=(0.5, 0.999))
optimizer_D = optim.Adam(discriminator.parameters(), lr=0.0002, betas=(0.5, 0.999))

# Step 10 — Train the GAN
# Cell 18 of 30
# Step 10 — Train the GAN
# The Discriminator learns from real images and detached fake images. The Generator then learns to make fake images that the Discriminator classifies as real.

# Code cell 19
# Cell 19 of 30
num_epochs = 5
generator_losses = []
discriminator_losses = []

for epoch in range(num_epochs):
    running_g_loss = 0.0
    running_d_loss = 0.0

    for real_images, _ in train_loader:
        real_images = real_images.to(device)
        current_batch_size = real_images.size(0)

        real_labels = torch.ones(current_batch_size, 1, device=device)
        fake_labels = torch.zeros(current_batch_size, 1, device=device)

        # Train Discriminator
        optimizer_D.zero_grad()

        real_output = discriminator(real_images)
        loss_real = criterion(real_output, real_labels)

        noise = torch.randn(current_batch_size, latent_dim, device=device)
        fake_images = generator(noise)
        fake_output = discriminator(fake_images.detach())
        loss_fake = criterion(fake_output, fake_labels)

        loss_D = loss_real + loss_fake
        loss_D.backward()
        optimizer_D.step()

        # Train Generator
        optimizer_G.zero_grad()

        noise = torch.randn(current_batch_size, latent_dim, device=device)
        generated_images = generator(noise)
        output = discriminator(generated_images)
        loss_G = criterion(output, real_labels)

        loss_G.backward()
        optimizer_G.step()

        running_g_loss += loss_G.item()
        running_d_loss += loss_D.item()

    avg_g_loss = running_g_loss / len(train_loader)
    avg_d_loss = running_d_loss / len(train_loader)
    generator_losses.append(avg_g_loss)
    discriminator_losses.append(avg_d_loss)

    print(
        f"Epoch [{epoch + 1}/{num_epochs}] | "
        f"Generator Loss: {avg_g_loss:.4f} | "
        f"Discriminator Loss: {avg_d_loss:.4f}"
    )

# Step 11 — Plot training losses
# Cell 20 of 30
# Step 11 — Plot training losses
# The two losses can change during training as the Generator and Discriminator compete. A GAN's losses do not necessarily decrease smoothly.

# Code cell 21
# Cell 21 of 30
plt.figure(figsize=(8, 5))
plt.plot(range(1, num_epochs + 1), generator_losses, marker="o", label="Generator")
plt.plot(range(1, num_epochs + 1), discriminator_losses, marker="o", label="Discriminator")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("GAN Training Loss")
plt.legend()
plt.grid(True)
plt.show()

# Step 12 — Generate new images
# Cell 22 of 30
# Step 12 — Generate new images
# Use random noise to create 16 new images after training.

# Code cell 23
# Cell 23 of 30
generator.eval()

with torch.no_grad():
    fixed_noise = torch.randn(16, latent_dim, device=device)
    generated_images = generator(fixed_noise).cpu()

plt.figure(figsize=(6, 6))
for i in range(16):
    plt.subplot(4, 4, i + 1)
    plt.imshow(generated_images[i].squeeze(), cmap="gray", vmin=-1, vmax=1)
    plt.axis("off")
plt.suptitle("GAN Generated Images")
plt.tight_layout()
plt.show()

# Step 13 — Compare real and generated images
# Cell 24 of 30
# Step 13 — Compare real and generated images
# The top row contains real MNIST images. The bottom row contains images created by the Generator.

# Code cell 25
# Cell 25 of 30
real_batch, _ = next(iter(train_loader))

plt.figure(figsize=(10, 4))
for i in range(5):
    plt.subplot(2, 5, i + 1)
    plt.imshow(real_batch[i].squeeze(), cmap="gray", vmin=-1, vmax=1)
    plt.title("Real")
    plt.axis("off")

    plt.subplot(2, 5, i + 6)
    plt.imshow(generated_images[i].squeeze(), cmap="gray", vmin=-1, vmax=1)
    plt.title("Generated")
    plt.axis("off")

plt.tight_layout()
plt.show()

# Step 14 — Save the generated images
# Cell 26 of 30
# Step 14 — Save the generated images
# Save the output grid as a PNG file.

# Code cell 27
# Cell 27 of 30
output_file = "gan_generated_images.png"

plt.figure(figsize=(6, 6))
for i in range(16):
    plt.subplot(4, 4, i + 1)
    plt.imshow(generated_images[i].squeeze(), cmap="gray", vmin=-1, vmax=1)
    plt.axis("off")
plt.tight_layout()
plt.savefig(output_file, dpi=150, bbox_inches="tight")
plt.show()

print("Saved:", output_file)

# Step 15 — Download the generated image
# Cell 28 of 30
# Step 15 — Download the generated image
# Run this cell in Google Colab to download the PNG file to your computer.

# Code cell 29
# Cell 29 of 30
from google.colab import files
files.download("gan_generated_images.png")
```
